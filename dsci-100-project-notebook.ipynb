#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

- Use `altair` to plot the distribution of one of the numeric variables in your tidied data set.

- Use `altair` to create a visualization showing the relative proportions of each category/level for one of categorical variable in your tidied data set. 

- Clearly label the variable that is being plotted in each plot.

*Notes:* 
- *Be sure the output is as clear as possible, provide an English title along with readable x and y column names*

- *You may also add more code cells to this section*

</div>

```python
import altair as alt

reg_wins_plot = alt.Chart(march_madness_tidy).mark_bar().encode(
    x=alt.X('reg_wins_pct:Q').bin(True).title('Regional Win Percentage'),
    y=alt.Y('count():Q').title('Number of Teams')
).properties(title='Distribution of Regional Wins Percentages')

reg_wins_plot
```

```python
proportions = (
    march_madness_tidy.groupby('tourney_finish')
    .size()
    .reset_index(name='count')
)

proportions['proportion'] = proportions['count'] / proportions['count'].sum()

proportions_plot = alt.Chart(proportions).mark_bar().encode(
    x=alt.X('tourney_finish:N').title('Tournament Finish').sort('-y').axis(labelAngle=-45),
    y=alt.Y('proportion:Q').title('Proportion of Teams').axis(format='%')
).properties(title='Proportion of Teams by Tournament Finish', width=350, height=400)

proportions_plot
```

Visualization 1: This plot compares regional win percentages across the tournament finishing categories (i.e. finishing position). If the distributions of regional win percentages differ between categories, it could be a good predictor in the classification model. 

```python
win_finish_plot = alt.Chart(summary).mark_bar().encode(
    x=alt.X('tourney_finish:N').title('Tournament Finish').axis(labelAngle=-45),
    y=alt.Y('reg_wins_pct:Q').title('Regional Win Percentage')
).properties(title='Regional Win Percentage by Tournament Finish', width=450, height=300)

win_finish_plot
```

Visualization 2: This plot compares tournament finishing positions for teams that played their first game at home and teams that didn't. It shows whether the distribution of finishing positions differ between them. This way, if there is a clear difference, that could indicate something about the relationship between playing the first game at home and the finishing position of the team. 

```python
home_finish_plot = alt.Chart(march_madness_tidy).mark_bar().encode(
    x=alt.X('first_game_at_home:N').title('First Tournament Game at Home').axis(labelAngle=0),
    y=alt.Y('count():Q').stack('normalize').title('Proportion of Teams').axis(format='%'),
    color=alt.Color('tourney_finish:N').title('Tournament Finish')
).properties(title='Tournament Finish by If First Game Was Played Home', width=100, height=400)

home_finish_plot
```
